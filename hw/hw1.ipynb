```python
# ANA670 - Week 1 HW - Q1
# define objective: `a square encloses the greatest area among all rectangles`
# define constraints: `assuming the total length of the four edges is fixed`

# constraints for rectangle: P(perimeter) = 2x + 2y
# y = P/2 - x

# define objective in terms of x.
# A(area) = x * y
# A = x * (P/2 - x)
# A = -x^2 + (P/2)x

# The result is a quadratic equation that can be solved where the maximum area is 
# the highest point of the parabola (which is upside down since x^2 is negative.

# A = ax^2 + bx + c
# A = -1x^2 + (P/2)x + 0
# a = -1, b = (P/2)

# The maximum of a quadratic equation is:
# x = -b/2a
# x = -(P/2)/2(-1)
# x = P/4

# Since the maximum of the curve is P/4, which is four equal sides, the proof shows that
# a square encloses the greatest area among all rectangles.

import matplotlib.pyplot as plt
import numpy as np

P = 20
x = np.linspace(0, P/2, 10000)
A = -x**2 + P/2 * x

plt.plot(x, A, color="blue", linewidth=2)
plt.title("Rectangle Area Parabola")
plt.xlabel("Width (x)")
plt.ylabel("Area (A)")
plt.grid(True)
plt.show()

max = np.argmax(A)
print(f'Max x = {x[max]:.2f}')
print(f'Max y = {(P / 2) - x[max]:.2f}')
print(f'Max A = {A[max]:.2f}')
```

```python
# ANA670 - Week 1 HW - Q2
# define objective: `a cylinder encloses the greatest volume with minimal material (surface area)`
# define constraints: size of radius (r) and height (h)

# Volume (V) of a cylinder: V = pi * r^2 * h
# Surface area (A) of a cylinder: A = 2 * pi * r^2 + 2 * pi * r * h

# Volume in terms of height (h): h = V / (pi * r^2)
# Surface area in terms of radius (r): A = 2 * pi * r^2 + 2 * pi * r * (V / pi * r^2)
# A = 2 * pi * r^2 + 2V / r

# Take the derivative to get the slope of the curve:
# dA/dr = 4 * pi * r - 2V / r^2

# Set the derivative to zero to get the minimum material/surface area
# 4 * pi * r - 2V / r^2 = 0
# 4 * pi * r = 2V / r^2
# 2 * pi * r^3 = V

# Substitue volume equation for V in the optimal volume expression from above
# 2 * pi * r^3 = pi * r^2 * h
# 2r = h

# 2r is just the diameter so the relationship of optimal volume for r and h
# is when the diameter (2r) is the same as the height (h).

V = 100
r = np.linspace(1.5, 5, 10000)
A = 2 * np.pi * r**2 + (2 * V) / r

plt.plot(r, A, color="blue", linewidth=2)
plt.title("Cylinder Surface Area Curve")
plt.xlabel("Radius (r)")
plt.ylabel("Material Area (A)")
plt.grid(True)
plt.show()

min_idx = np.argmin(A)
print(f'Min r = {r[min_idx]:.2f}')
print(f'Min h = {V / (np.pi * r[min_idx]**2):.2f}')
print(f'Min A = {A[min_idx]:.2f}')
```

```python
# ANA670 - Week 1 HW - Q4
# Find the global maximum for linear optimization 
# f(x, y) = x+y 
# subject to 0 < x < 5, 0 < y < 2 and 2x + y = 8

# Get all equations in terms of x:
# 2x + y = 8
# y = -2x + 8

# account for y constraints:
# 0 < -2x + 8 < 2
# 3 < x < 4 (new full x,y constraints 
# because 3 < x < 4 falls within 0 < x < 5 and is more restrictive)

# substitute constraints back into main equation
# f(x) = x + y
# f(x) = x + -2x + 8
# f(x) = -x + 8

# This is a linear equation with a (-1) slope. That means the max value
# is the farthest we can go left within the constraints. At x = 3 we get a result of 5.
# Since the constraint is x > 3, we can never actually get to 3. We can have 3.01,
# 3.001 and keep going and adding infinite zeros without actually getting to 3. 
# Because of this, the max value does not exist and there is no global max within the constraints.

x = np.linspace(2, 5, 100)
x_allowed = np.linspace(3, 4, 100)

plt.plot(x, -x + 8, "g--", label="f(x) = -x + 8")
plt.plot(x_allowed, -x_allowed + 8, "b-", linewidth=3, label="3 < x < 4")

plt.title("f(x) = -x + 8")
plt.grid(True)
plt.legend()
plt.show()
```

```python
# ANA670 - Week 1 HW - Q6
from matplotlib import cm

def compute_hessian(f, eq_string, x=0.0, y=1.0, h=1e-5, graph=True):
    df_dx = lambda x, y: (f(x + h, y) - f(x - h, y)) / (2 * h)
    df_dy = lambda x, y: (f(x, y + h) - f(x, y - h)) / (2 * h)

    f_xx = (df_dx(x + h, y) - df_dx(x - h, y)) / (2 * h)
    f_yy = (df_dy(x, y + h) - df_dy(x, y - h)) / (2 * h)
    f_xy = (df_dx(x, y + h) - df_dx(x, y - h)) / (2 * h)
    f_yx = (df_dy(x + h, y) - df_dy(x - h, y)) / (2 * h) 

    matrix = np.round(np.array([[f_xx, f_xy], [f_yx, f_yy]]), decimals=4)
    if graph:
        fig = plt.figure(figsize=(6, 4))
        ax = fig.add_subplot(111, projection='3d')
        
        _x = np.linspace(-2, 2, 40)
        _y = np.linspace(0.1, 2, 40)
        X, Y = np.meshgrid(_x, _y)
        Z = np.vectorize(f)(X, Y)
        
        if "y" in eq_string and "/" in eq_string:
            Z = np.clip(Z, 0, 8)
            
        ax.plot_surface(X, Y, Z, cmap=cm.coolwarm, edgecolor='none', alpha=0.85)
        ax.set_title(f"3D Surface: f(x,y) = {eq_string}", fontsize=11, fontweight='bold')
        ax.set_xlabel('X')
        ax.set_ylabel('Y')
        ax.set_zlabel('Z')
        plt.show()
    return matrix

equations = [
    ("a", lambda x, y: x**2 + np.exp(y), "x² + exp(y)"),
    ("b", lambda x, y: 1 - x**2 - y**2, "1 - x² - y²"),
    ("c", lambda x, y: x * y,           "xy"),
    ("d", lambda x, y: (x**2) / y if y != 0 else 0, "x² / y"),
]

for label, eq, eq_string in equations:
    print(f"{label}) f(x,y) = {eq_string}")
    print(compute_hessian(eq, eq_string, x=0.0, y=1.0, graph=True))
    print()

print(f'a) is Convex because all diagonal curvatures are positive.')
print(f'b) is Concave because all diagonal curvatures are negative.')
print(f'c) is Neither because there is a saddle point.')
print(f'd) is Convex because the matrix is positive where y > 0.')
```

```python
# ANA670 - Week 1 HW - Q7

# (minima listed below). There are 1 to multiple minima based on the value of k.
# This is because x^2 wants to create a bowl shaped parabola and 
# cos^2(x) wants to create a wave. The larger the value of k, the more 
# amplified the wave across the curve which creates more local minima.

def f(x, k):return x**2 + k * np.cos(x)**2

x_vals = np.linspace(-30, 30, 6001)
ks = [1,2,3, 10]

plt.figure(figsize=(12, 6))

for k in ks:
    y_vals = f(x_vals, k)
    
    plt.plot(x_vals, y_vals, label=f'k = {k}', lw=2)
    minima_indices = np.where((y_vals[1:-1] < y_vals[:-2]) & (y_vals[1:-1] < y_vals[2:]))[0] + 1
    minima_x = x_vals[minima_indices]
    minima_y = y_vals[minima_indices]
    plt.scatter(minima_x, minima_y, color='red', zorder=5, s=50)

    print(f"Minima for k = {k}")
    print(f"Total minima found: {len(minima_x)}")
    for idx in range(len(minima_x)):
        print(f"  Minimum {idx+1}: x = {minima_x[idx]:.4f}, f(x) = {minima_y[idx]:.4f}")
    print()

plt.scatter([], [], color='red', s=50, label='Local Minima')
plt.title(r"Plot of $f(x) = x^2 + k\cos^2(x)$")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.xlim(-10, 10)  
plt.ylim(-1, 40)
plt.grid(True, linestyle="--", alpha=0.5)
plt.legend()

print(f'')
plt.show()
```
