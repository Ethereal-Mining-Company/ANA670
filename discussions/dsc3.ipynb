```python
import numpy as np
import matplotlib.pyplot as plt

def blend(alpha, link_multiplier, baseline_unit):
    return alpha * link_multiplier + (1 - alpha) * baseline_unit

# --- DYNAMIC CONFIGURATION BLOCK ---
baseline_unit = 1.0
alpha_marker = 0.85
points_alpha = np.array([0.0, 0.25, 0.5, 0.75, 0.85, 1.0])

# Central List: (Label, Multiplier, Color, Base Line Width, Line Style)
scenarios = [
    ('Concentrated Link', 5.0, 'teal', 1.5, '-'),
    ('Average Link (Baseline Benchmark)', 1.0, 'orange', 1.5, ':'),
    ('Diluted Link', 0.2, 'crimson', 3.0, '-')
]
# ------------------------------------

alpha = np.linspace(0, 1, 100)
plt.figure(figsize=(10, 6))

max_multiplier = max([item[1] for item in scenarios])
min_multiplier = min([item[1] for item in scenarios])
global_range = max_multiplier - min_multiplier
label_offset_distance = global_range * 0.03  

for name, multiplier, color, lw, ls in scenarios:
    prob_path = blend(alpha, multiplier, baseline_unit)
    plt.plot(alpha, prob_path, color=color, lw=lw, linestyle=ls, label=name)
    pts_y = blend(points_alpha, multiplier, baseline_unit)
    dot_size = lw * 25
    plt.scatter(points_alpha, pts_y, color=color, edgecolor='black', s=dot_size, zorder=5)
    v_align = 'bottom' if multiplier >= baseline_unit else 'top'
    offset = label_offset_distance if multiplier >= baseline_unit else -label_offset_distance
    
    for a, y_val in zip(points_alpha, pts_y):
        plt.text(a, y_val + offset, f"{y_val:.2f}x", ha='center', va=v_align, 
                 fontsize=9, fontweight='bold', color=color)

plt.axvline(x=alpha_marker, color='gray', linestyle='--', alpha=0.5)
plt.title("Impact of Alpha on Concentrated vs. Diluted Links", fontsize=14, fontweight='bold', pad=15)
plt.xlabel("Alpha (α)")
plt.ylabel("Relative Power Multiplier")

plt.xlim(min(points_alpha) - 0.05, max(points_alpha) + 0.05)
plt.ylim(min_multiplier - (global_range * 0.15), max_multiplier + (global_range * 0.15))

plt.grid(True, linestyle=':', alpha=0.6)
plt.legend(loc='upper left')
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# --- 100% UNIVERSAL INVARIANT CONFIGURATION DASHBOARD ---
total_executions = 25  
# DYNAMIC: You can add, remove, or modify ANY number of alphas here!
alphas_to_test = [0.0, 0.15, 0.85, 1.0]  

scenarios = [
    ('Concentrated Link Type',            5.0, 'teal',    1.5, '-', 0.01),
    ('Average Link (Baseline Benchmark)', 1.0, 'orange',  2.0, ':', 0.29),
    ('Diluted Link Type',                 0.2, 'crimson', 2.0, '-', 0.70)
]
# --------------------------------------------------------

num_groups = len(scenarios)
group_fractions = np.array([row[5] for row in scenarios], dtype=float)
multipliers = np.array([row[1] for row in scenarios], dtype=float)

# Base Assemblies (Shared blueprint constants)
group_weights = multipliers * group_fractions
global_distribution = group_weights / np.sum(group_weights)

M_macro = np.zeros((num_groups, num_groups))
for j in range(num_groups):
    M_macro[:, j] = 0.1 * global_distribution
    M_macro[j, j] += 0.9  

E_macro = np.vstack([group_fractions, group_fractions, group_fractions]).T

print("=" * 80)
print("             SECTION 1: THE ECOSYSTEM COMPONENT DESIGN BLUEPRINT            ")
print("=" * 80)
print("METADATA PARAMETERS:")
print(f"  - Target Iteration Horizon Limit:             {total_executions} steps")
print(f"  - Network Category Profile Count:             {num_groups} macro types\n")

print("CONFIGURED SCENARIOS CATEGORIES:")
for label, mult, color, lw, ls, frac in scenarios:
    print(f"  * {label:<35} | Target: {mult:3.1f}x | Share: {frac*100:4.1f}%")
print("")

print(f"-> 1. EXTRACTED POPULATION FRACTIONS (Web Land Share):\n   {group_fractions}")
print(f"-> 2. GLOBAL SYSTEM MULTIPLIER LINK WEIGHT VECTOR:\n   {global_distribution}\n")
print(f"-> 3. ORGANIC LINK MATRIX (M_macro with 90% friction loop traps):\n{M_macro}\n")
print(f"-> 4. RANDOM TELEPORTATION TARGET MATRIX (E_macro):\n{E_macro}\n")

# --- INITIALIZE MULTI-GRAPH CANVAS ---
fig, axes = plt.subplots(len(alphas_to_test), 1, figsize=(11, 4 * len(alphas_to_test)), sharex=True)

if len(alphas_to_test) == 1:
    axes = [axes]

print("=" * 80)
print("             SECTION 2: MULTI-ALPHA EXECUTION AUDIT LOG                     ")
print("=" * 80)

# --- THE DYNAMIC MULTI-ALPHA EXECUTION ENGINE ---
for ax_idx, alpha in enumerate(alphas_to_test):
    # Calculate the precise Google Matrix for this current alpha loop
    google_matrix = alpha * M_macro + (1 - alpha) * E_macro
    
    # Initialize variables for this specific alpha calculation
    current_multipliers = np.ones(num_groups)
    multiplier_history = np.zeros((total_executions + 1, num_groups))
    multiplier_history[0, :] = current_multipliers
    
    # Run the full Power Iteration clicks loop (Calculations handled silently)
    for step in range(1, total_executions + 1):
        group_shares = current_multipliers * group_fractions
        next_group_shares = google_matrix.dot(group_shares)
        current_multipliers = next_group_shares / group_fractions
        multiplier_history[step, :] = current_multipliers
        
    # Print the core matrix architecture and the final output summary for this specific Alpha condition
    print(f"\n[ CONDITIONAL CONVERGENCE PROFILE FOR α = {alpha} ]")
    print("-" * 55)
    print(f"-> 1. THE GOOGLE TRANSITION ENGINE MATRIX (G):\n{google_matrix}\n")
    print(f"-> 2. PROBABILITY CONSERVATION COLUMN SUM CHECK: {np.sum(google_matrix, axis=0)}")
    print(f"-> 3. INITIAL STATE MULTIPLIERS (Step 00):       {multiplier_history[0, :]}")
    print(f"-> 4. FINAL CONVERGED MULTIPLIERS   (Step {total_executions:02d}):       {multiplier_history[-1, :]}")
    print("." * 80)
        
    # Formatting and graph boundaries calculation for this specific subplot
    max_val = np.max(multiplier_history)
    min_val = np.min(multiplier_history)
    global_range = max_val - min_val
    label_offset = global_range * 0.04
    
    execution_steps = np.arange(total_executions + 1)
    points_step = np.linspace(0, total_executions, 6, dtype=int)
    
    # Plot each line track inside this specific alpha's axis pane
    for index, (name, _, color, lw, ls, _) in enumerate(scenarios):
        page_multiplier_history = multiplier_history[:, index]
        
        axes[ax_idx].plot(execution_steps, page_multiplier_history, color=color, lw=lw, linestyle=ls, label=name)
        axes[ax_idx].scatter(points_step, page_multiplier_history[points_step], color=color, edgecolor='black', s=40, zorder=5)
        
        v_align = 'bottom' if index == 0 else 'top'
        offset = label_offset if index == 0 else -label_offset
        
        for x in points_step:
            y_val = page_multiplier_history[x]
            axes[ax_idx].text(x, y_val + offset, f"{y_val:.2f}x", ha='center', va=v_align, 
                               fontsize=8, fontweight='bold', color=color)
            
    # Final styling polish for this specific chart slice
    axes[ax_idx].set_title(f"Convergence Trajectory under Alpha Condition (α = {alpha})", fontsize=11, fontweight='bold')
    axes[ax_idx].set_ylabel("Relative Power Multiplier", fontsize=10)
    axes[ax_idx].grid(True, linestyle=':', alpha=0.6)
    axes[ax_idx].set_xlim(-0.5, total_executions + 0.5)
    axes[ax_idx].set_ylim(min_val - (global_range * 0.20), max_val + (global_range * 0.20))
    axes[ax_idx].legend(loc='upper right', fontsize=8)

# Label the shared bottom horizontal timeline axis
axes[-1].set_xlabel("Execution Step (Iteration Index)", fontsize=11)
plt.xticks(execution_steps)

plt.tight_layout()
print("\n" + "="*80 + "\nAll loops processed. Displaying timeline charts below...")
plt.show()
```
